```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load data from csv file
data_path = Path("kimchi_playtime.csv")
kimchi_data = pd.read_csv(data_path)
kimchi_data.head()

#plot playtime as a line grpah with error bars for mean and std dev
plt.figure(figsize=(12, 6))
sns.lineplot(data=kimchi_data, x='Time of Day', y='Mean times played', marker='o', color='blue')
plt.errorbar(
    kimchi_data['Time of Day'],
    kimchi_data['Mean times played'],
    yerr=kimchi_data['SD'],
    fmt='o-',
    capsize=4, 
    ecolor='orange'
)
plt.title('Kimchi Playtime Over Time')
plt.xlabel('Date')
plt.ylabel('Playtime')

#do the same as a bar graph
plt.figure(figsize=(12, 6))
sns.barplot(data=kimchi_data, x='Time of Day', y='Mean times played', color='lightblue')
plt.errorbar(
    kimchi_data['Time of Day'],
    kimchi_data['Mean times played'],
    yerr=kimchi_data['SD'],
    fmt='o',
    capsize=4, 
    ecolor='orange'
)
#change time of day to date format
kimchi_data['Time of Day'] = pd.to_datetime(kimchi_data['Time of Day'])
#change time of day to hours
kimchi_data['Time of Day'] = kimchi_data['Time of Day'].dt.hours
plt.title('Kimchi Playtime Over Time')
plt.xlabel('Date')
plt.ylabel('Playtime')
plt.show()
```
